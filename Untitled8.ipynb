```python
from google.colab import drive
drive.mount('/content/drive')

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('/content/drive/MyDrive/hearing_test.csv')
print(df.shape)
df.head()
```

```python
df.info()
df.describe().transpose()

# Visualize the relationship
sns.scatterplot(data=df, x='age', y='physical_score', hue='test_result', alpha=0.3)
plt.title("Age vs Physical Score — colored by test result")
plt.show()

print("Class distribution:")
print(df['test_result'].value_counts())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X = df.drop('test_result', axis=1)   # features: age, physical_score
y = df['test_result']                 # target: 0 or 1

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.1, random_state=101
)

# Scale — Logistic Regression uses gradient descent, scaling is important
scaler = StandardScaler()
scaled_X_train = scaler.fit_transform(X_train)  # fit AND transform training
scaled_X_test  = scaler.transform(X_test)        # ONLY transform test (no fit!)

print("Train size:", X_train.shape)
print("Test size: ", X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

log_model = LogisticRegression()
log_model.fit(scaled_X_train, y_train)

print("Model trained!")
print("Coefficients:", log_model.coef_)      # weights for each feature
print("Intercept:", log_model.intercept_)     # bias term
```

```python
from sklearn.metrics import (accuracy_score, confusion_matrix,
                              classification_report, ConfusionMatrixDisplay)

y_pred = log_model.predict(scaled_X_test)

# Accuracy
print("Accuracy:", accuracy_score(y_test, y_pred))

# Confusion Matrix
ConfusionMatrixDisplay.from_predictions(y_test, y_pred, normalize='true')
plt.title("Confusion Matrix — Hearing Test")
plt.show()

# Full Report
print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import roc_curve, roc_auc_score

# Get probabilities (not just 0/1 predictions)
y_prob = log_model.predict_proba(scaled_X_test)[:, 1]
# [:,1] gets the probability of class 1 (pass) for each sample

# Calculate ROC curve points
fpr, tpr, thresholds = roc_curve(y_test, y_prob)

# Plot
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, label=f"Logistic Regression (AUC = {roc_auc_score(y_test, y_prob):.3f})")
plt.plot([0,1], [0,1], 'k--', label='Random Guessing (AUC = 0.5)')
plt.xlabel("False Positive Rate (FPR)")
plt.ylabel("True Positive Rate (TPR = Recall)")
plt.title("ROC Curve — Hearing Test")
plt.legend()
plt.grid()
plt.show()

print("AUC Score:", roc_auc_score(y_test, y_prob))
```
